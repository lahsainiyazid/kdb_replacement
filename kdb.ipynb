```python
features = [
    "Goals p 90",
    "Assists p 90",
    "Expected Goals",
    "Shots p 90",
    "Key passes",
    "Shot creating actions p 90",
    "Goal creating actions p 90",
    "Pass completion %",
    "Progressive Passes",
    "Passes into penalty area",
    "Progressive Carries",
    "carries final 3rd",
    "% Successful take-ons",
    "Tackles Won",
    "Interceptions",
    "Passes blocked",
    "% Aerial Duels won"
]
```

```python
from sklearn.preprocessing import StandardScaler 
X=df[features]
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
from sklearn.neighbors import NearestNeighbors
knn=NearestNeighbors(n_neighbors=11,metric="euclidean")
knn.fit(X_scaled)
```

```python
kdb_idx=(df[df["player"]=="Kevin De Bruyne"]).index[0]
kdb_idx
```

```python
kdb_pos=df.index.get_loc(kdb_idx)
kdb_pos
```

```python
kdb_vector=X_scaled[kdb_pos]
kdb_vector=kdb_vector.reshape(1,-1)
```

```python
distances,indices=knn.kneighbors(kdb_vector,n_neighbors=11)
neighbors=df.iloc[indices[0]]
replacements=neighbors[(neighbors['age']<=21) & ((neighbors['pos']=='MF')|(neighbors['pos']=='FW'))]
replacements
```
